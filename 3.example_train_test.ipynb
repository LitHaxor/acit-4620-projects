```python
import torch.nn as nn
import torch.nn.functional as F

class HazeFilters(nn.Module):
    # turns an RGB image (3 channels) into RGB + 5 haze filter maps (8 channels)
    def forward(self, rgb):
        gray = rgb.mean(dim=1, keepdim=True)
        brightest = rgb.max(dim=1, keepdim=True).values
        darkest = rgb.min(dim=1, keepdim=True).values

        dark_channel = -F.max_pool2d(-darkest, kernel_size=7, stride=1, padding=3)
        saturation = (brightest - darkest) / (brightest + 1e-6)
        brightness = brightest

        sobel_x = torch.tensor([[-1., 0., 1.], [-2., 0., 2.], [-1., 0., 1.]], device=rgb.device).view(1, 1, 3, 3)
        edges = torch.sqrt(F.conv2d(gray, sobel_x, padding=1) ** 2 + F.conv2d(gray, sobel_x.transpose(2, 3), padding=1) ** 2)

        mean = F.avg_pool2d(gray, 7, stride=1, padding=3)
        mean_of_squares = F.avg_pool2d(gray ** 2, 7, stride=1, padding=3)
        local_contrast = (mean_of_squares - mean ** 2).clamp(min=0).sqrt()

        return torch.cat([rgb, dark_channel, saturation, brightness, edges, local_contrast], dim=1)
```

```python
import torch.nn as nn

model = nn.Sequential(
    HazeFilters(),  # 3 x 144 x 192 -> 8 x 144 x 192
    # Stem: 8 x 144 x 192 -> 32 x 72 x 96
    nn.Conv2d(
        8, 32,         # 8 input channels: RGB + 5 haze filter maps
        kernel_size=5, # bigger 5x5 filter to see more of the image at once
        stride=2,      # jumps 2 pixels at a time -> halves the size, makes training faster
        padding=2
    ),
    nn.BatchNorm2d(32), # keeps the numbers in a stable range -> steadier learning
    nn.ReLU(),

    # Block 1: 32 x 72 x 96 -> 64 x 36 x 48
    nn.Conv2d(32, 64, kernel_size=3, padding=1),
    nn.BatchNorm2d(64),
    nn.ReLU(),
    nn.Conv2d(64, 64, kernel_size=3, padding=1), # 2 convolutions before pooling -> learns richer patterns
    nn.BatchNorm2d(64),
    nn.ReLU(),
    nn.MaxPool2d(2),

    # Block 2: 64 x 36 x 48 -> 128 x 18 x 24
    nn.Conv2d(64, 128, kernel_size=3, padding=1),
    nn.BatchNorm2d(128),
    nn.ReLU(),
    nn.Conv2d(128, 128, kernel_size=3, padding=1),
    nn.BatchNorm2d(128),
    nn.ReLU(),
    nn.MaxPool2d(2),

    # Block 3: 128 x 18 x 24 -> 256 x 9 x 12
    nn.Conv2d(128, 256, kernel_size=3, padding=1),
    nn.BatchNorm2d(256),
    nn.ReLU(),
    nn.MaxPool2d(2),

    nn.AdaptiveAvgPool2d(1), # Global Average Pooling -> 256 x 1 x 1
    nn.Flatten(),            # -> 256 numbers

    # Head: combine the 256 features into one AQI value
    nn.Linear(256, 64),
    nn.ReLU(),
    nn.Dropout(0.3), # switches off 30% of the values while training -> less memorising
    nn.Linear(64, 1) # Output Layer: predicted AQI / 100
)
```

```python
"""
[3, 144, 192]    photo (3 colours)
[16, 72, 96]     after block 1  (16 maps, half size)
[32, 36, 48]     after block 2
[64, 18, 24]     after block 3
[128, 18, 24]    after the last convolution
[128]            after averaging + flatten
[1]              predicted AQI
"""
```

```python

from torch.utils.data import TensorDataset, DataLoader

device_type = ''

if torch.backends.mps.is_available():
    device_type = 'mps'
elif torch.backends.cuda.is_available():
    device_type = 'cuda'
else:
    device_type = 'cpu'

device = torch.device(device_type)

device
```

```python
model = model.to(device)


model
```

```python
learn_loader = DataLoader(TensorDataset(learn_images, learn_aqi), batch_size=64, shuffle=True)
val_loader = DataLoader(TensorDataset(val_images, val_aqi), batch_size=256, shuffle=False)


learn_loader, val_loader
```

```python
loss_function = nn.L1Loss()
loss_function
```

```python
optimizer = torch.optim.Adam(model.parameters(), lr=0.003,weight_decay=0.0001)

optimizer
```

```python
baseline_mae = (val_aqi - learn_aqi.median()).abs().mean().item()
print('validation MAE if we always guess the median AQI:', round(baseline_mae, 1))
```

```python
def predict(model: nn.Module, loader: DataLoader) -> torch.Tensor:
    model.eval()
    predictions = []

    with torch.no_grad():
        for images, aqi in loader:
            output = model(images.to(device)).squeeze(1)  # [batch, 1] -> [batch]
            predictions.append(output.cpu() * 100)        # back from AQI/100 to AQI

    return torch.cat(predictions)
```

```python
def augment(images):
    # 1) mirror about half of the photos left-right
    flip = torch.rand(images.shape[0], device=images.device) < 0.5
    images[flip] = images[flip].flip(3)

    # 2) shift the batch a few pixels: up to 8 left/right and 4 up/down
    padded = F.pad(images, (8, 8, 4, 4), mode='replicate')
    x = int(torch.randint(0, 17, (1,)))
    y = int(torch.randint(0, 9, (1,)))
    images = padded[:, :, y:y + 144, x:x + 192]

    return images
```

```python
import copy

checkpoints = {
    'MAE': [],
    'RMSE': []
}
epochs = 10
best_mae = float('inf')  # best validation MAE so far (starts as "infinitely bad")
best_epoch = 0

for epoch in range(epochs):
    model.train()
    for images, aqi in tqdm(learn_loader, desc=f'epoch {epoch + 1}', leave=False):
        images = images.to(device)
        images = augment(images)
        target = (aqi / 100).to(device)

        prediction = model(images).squeeze(1)
        loss = loss_function(prediction, target)

        optimizer.zero_grad()
        loss.backward()
        optimizer.step()

    errors = predict(model, val_loader) - val_aqi  # one error per validation photo
    checkpoints['MAE'].append(errors.abs().mean().item())
    checkpoints['RMSE'].append((errors ** 2).mean().sqrt().item())
    print(f"epoch {epoch + 1}: validation MAE {checkpoints['MAE'][-1]:.1f} | RMSE {checkpoints['RMSE'][-1]:.1f}")

    # keep a copy of the weights whenever the validation MAE is the best so far
    if checkpoints['MAE'][-1] < best_mae:
        best_mae = checkpoints['MAE'][-1]
        best_epoch = epoch + 1
        best_weights = copy.deepcopy(model.state_dict())

# put the best weights back into the model and save them to a file
model.load_state_dict(best_weights)
torch.save(best_weights, 'best_model.pt')
print(f'best model: epoch {best_epoch} with validation MAE {best_mae:.1f} (saved to best_model.pt)')
```

```python
history = pd.DataFrame(checkpoints)
history.index = history.index + 1

plt.figure(figsize=(8, 4))
plt.plot(history['MAE'], marker='o', label='MAE')
plt.plot(history['RMSE'], marker='o', label='RMSE')
plt.axhline(baseline_mae, color='gray', linestyle='--', label='baseline MAE (always guess median)')
plt.xlabel('Epoch')
plt.ylabel('Error (AQI)')
plt.title('Validation error per epoch')
plt.legend()
plt.show()
```

```python
# load a random test image from every AQI bin and predict its AQI

bins = ['0–50', '51–100', '101–150', '151–200', '201–300', '301–600']
randomness = np.random.default_rng()

# load the best model weights from the file
best_weights = torch.load('best_model.pt')
model.load_state_dict(best_weights)

model.eval()
plt.figure(figsize=(12, 7))

for number, aqi_bin in enumerate(bins):
    bin_rows = test_sets[test_sets['pm25_bin'] == aqi_bin]
    row = bin_rows.iloc[randomness.integers(0, len(bin_rows))]

    image = Image.open(row['image_path']).convert('RGB').resize([192, 144])
    true_aqi = row['pm25']

    with torch.no_grad():
        predicted_aqi = model(load_image(row['image_path']).unsqueeze(0).to(device)).item() * 100

    error = abs(predicted_aqi - true_aqi)

    plt.subplot(2, 3, number + 1)
    plt.imshow(image)
    plt.title(f"Bin {aqi_bin}\nPredicted AQI: {predicted_aqi:.0f}\nActual AQI: {true_aqi:.0f}")
    plt.text(0, 0, f"Error: {error:.1f}", color='white', fontsize=12, bbox=dict(facecolor='black', alpha=0.5))
    plt.axis('off')

plt.tight_layout()
plt.show()
```

```python
from sklearn.model_selection import StratifiedGroupKFold

folds = StratifiedGroupKFold(n_splits=5, shuffle=True, random_state=42)

# y = the AQI bin   -> every fold gets the same share of each bin
# groups = station  -> all photos of one station stay in the same fold
splits = list(folds.split(train_sets, train_sets['pm25_bin'], groups=train_sets['station_id']))

[len(val_index) for learn_index, val_index in splits]
```

```python
# share of each AQI bin in every validation fold (every column should look almost the same)
share = {}
for fold, (learn_index, val_index) in enumerate(splits):
    share[f'fold {fold + 1}'] = train_sets['pm25_bin'].iloc[val_index].value_counts(normalize=True).reindex(bins)
share['all'] = train_sets['pm25_bin'].value_counts(normalize=True).reindex(bins)

pd.DataFrame(share).round(3)
```

```python
# train a fresh model on every fold and keep its best validation MAE
epochs = 10
fold_mae = []

for fold, (learn_index, val_index) in enumerate(splits):
    fold_learn_loader = DataLoader(TensorDataset(train_images[learn_index], train_aqi[learn_index]), batch_size=64, shuffle=True)
    fold_val_loader = DataLoader(TensorDataset(train_images[val_index], train_aqi[val_index]), batch_size=256, shuffle=False)
    fold_val_aqi = train_aqi[val_index]
    fold_baseline = (fold_val_aqi - train_aqi[learn_index].median()).abs().mean().item()

    # start every fold from new random weights and a new optimizer
    for layer in model.modules():
        if hasattr(layer, 'reset_parameters'):
            layer.reset_parameters()
    optimizer = torch.optim.Adam(model.parameters(), lr=0.003, weight_decay=0.0001)

    best_mae = float('inf')
    best_epoch = 0

    for epoch in range(epochs):
        model.train()
        for images, aqi in tqdm(fold_learn_loader, desc=f'fold {fold + 1} epoch {epoch + 1}', leave=False):
            images = images.to(device)
            images = augment(images)
            target = (aqi / 100).to(device)

            prediction = model(images).squeeze(1)
            loss = loss_function(prediction, target)

            optimizer.zero_grad()
            loss.backward()
            optimizer.step()

        mae = (predict(model, fold_val_loader) - fold_val_aqi).abs().mean().item()
        if mae < best_mae:
            best_mae = mae
            best_epoch = epoch + 1

    fold_mae.append(best_mae)
    print(f'fold {fold + 1}: best validation MAE {best_mae:.1f} (epoch {best_epoch}) | baseline {fold_baseline:.1f}')

print('MAE per fold:', [round(m, 1) for m in fold_mae])
print(f'average MAE: {np.mean(fold_mae):.1f} ± {np.std(fold_mae):.1f}')
```

```python
class SqueezeExcite(nn.Module):
    # learns one weight (0..1) per channel from the whole image, then scales each channel by it
    # -> the model can turn up the channels that carry haze information
    def __init__(self, channels, reduction=8):
        super().__init__()
        self.weights = nn.Sequential(
            nn.AdaptiveAvgPool2d(1),                      # one average per channel
            nn.Flatten(),
            nn.Linear(channels, channels // reduction),   # squeeze: 128 -> 16
            nn.ReLU(),
            nn.Linear(channels // reduction, channels),   # excite: 16 -> 128
            nn.Sigmoid(),                                 # weights between 0 and 1
        )

    def forward(self, x):
        return x * self.weights(x).view(x.shape[0], -1, 1, 1)


class ResidualBlock(nn.Module):
    # conv -> BN -> LeakyReLU (-> SE), add the input back (residual), then halve the size
    # adding the input back keeps earlier haze cues alive in deeper layers
    def __init__(self, channels, use_se=False):
        super().__init__()
        self.conv = nn.Sequential(
            nn.Conv2d(channels, channels, kernel_size=3, padding=1),
            nn.BatchNorm2d(channels),
            nn.LeakyReLU(0.1),
        )
        self.se = SqueezeExcite(channels) if use_se else nn.Identity()
        self.pool = nn.MaxPool2d(3, stride=2, padding=1)

    def forward(self, x):
        return self.pool(x + self.se(self.conv(x)))
```

```python
# ResSE model (Hasan & Shakir, 2026), trained from scratch, one output: AQI / 100
aqi_new_model = nn.Sequential(
    HazeFilters(),  # 3 x 144 x 192 -> 8 x 144 x 192

    # Block 1: 8 x 144 x 192 -> 64 x 72 x 96
    nn.Conv2d(8, 64, kernel_size=3, padding=1),
    nn.BatchNorm2d(64),
    nn.LeakyReLU(0.1),  # like ReLU, but lets small negative values through (x 0.1)
    nn.Conv2d(64, 64, kernel_size=3, padding=1),
    nn.BatchNorm2d(64),
    nn.LeakyReLU(0.1),
    nn.MaxPool2d(3, stride=2, padding=1),

    # Block 2: 64 x 72 x 96 -> 128 x 36 x 48
    nn.Conv2d(64, 128, kernel_size=3, padding=1),
    nn.BatchNorm2d(128),
    nn.LeakyReLU(0.1),
    nn.Conv2d(128, 128, kernel_size=3, padding=1),
    nn.BatchNorm2d(128),
    nn.LeakyReLU(0.1),
    nn.MaxPool2d(3, stride=2, padding=1),

    # Blocks 3-6: residual blocks, SE in block 3 and 6
    ResidualBlock(128, use_se=True),  # -> 128 x 18 x 24
    ResidualBlock(128),               # -> 128 x 9 x 12
    ResidualBlock(128),               # -> 128 x 5 x 6
    ResidualBlock(128, use_se=True),  # -> 128 x 3 x 3

    nn.AdaptiveAvgPool2d(1),  # Global Average Pooling -> 128 x 1 x 1
    nn.Flatten(),             # -> 128 numbers

    # Neck + head
    nn.Linear(128, 512),
    nn.LeakyReLU(0.1),
    nn.Dropout(0.3),
    nn.Linear(512, 512),
    nn.LeakyReLU(0.1),
    nn.Linear(512, 1)  # predicted AQI / 100
).to(device)

print('output shape:', aqi_new_model(learn_images[:4].to(device)).shape)
print('number of weights:', sum(p.numel() for p in aqi_new_model.parameters()))
```

```python
# same 5 folds for the ResSE model, with the paper's training recipe:
# small learning rate, halve it when validation stops improving, stop after 10 epochs without improvement
epochs = 20
new_fold_mae = []

for fold, (learn_index, val_index) in enumerate(splits):
    fold_learn_loader = DataLoader(TensorDataset(train_images[learn_index], train_aqi[learn_index]), batch_size=64, shuffle=True)
    fold_val_loader = DataLoader(TensorDataset(train_images[val_index], train_aqi[val_index]), batch_size=256, shuffle=False)
    fold_val_aqi = train_aqi[val_index]

    # start every fold from new random weights, a new optimizer and a new scheduler
    for layer in aqi_new_model.modules():
        if hasattr(layer, 'reset_parameters'):
            layer.reset_parameters()
    new_optimizer = torch.optim.Adam(aqi_new_model.parameters(), lr=0.0001)
    new_scheduler = torch.optim.lr_scheduler.ReduceLROnPlateau(new_optimizer, factor=0.5, patience=5, min_lr=1e-6)

    best_mae = float('inf')
    best_epoch = 0

    for epoch in range(epochs):
        aqi_new_model.train()
        for images, aqi in tqdm(fold_learn_loader, desc=f'fold {fold + 1} epoch {epoch + 1}', leave=False):
            images = images.to(device)
            images = augment(images)
            target = (aqi / 100).to(device)

            prediction = aqi_new_model(images).squeeze(1)
            loss = loss_function(prediction, target)

            new_optimizer.zero_grad()
            loss.backward()
            new_optimizer.step()

        mae = (predict(aqi_new_model, fold_val_loader) - fold_val_aqi).abs().mean().item()
        new_scheduler.step(mae)  # halves the learning rate after 5 epochs without improvement

        if mae < best_mae:
            best_mae = mae
            best_epoch = epoch + 1
        elif epoch + 1 - best_epoch >= 10:
            break  # early stopping: 10 epochs without improvement

    new_fold_mae.append(best_mae)
    print(f'fold {fold + 1}: best validation MAE {best_mae:.1f} (epoch {best_epoch})')

print('MAE per fold:', [round(m, 1) for m in new_fold_mae])
print(f'average MAE: {np.mean(new_fold_mae):.1f} ± {np.std(new_fold_mae):.1f}')
```
